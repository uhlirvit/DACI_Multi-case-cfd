```python
import pickle

# Load the file
with open('all_cases_data.pkl', 'rb') as f:
    data = pickle.load(f)

# Check structure
print(f"Cases loaded: {list(data.keys())}")
print(f"\nFirst case structure:")
case = data['U1_uniform']
print(f"  U_ref: {case['U_ref']}")
print(f"  Profile type: {case['profile_type']}")
print(f"  Number of profiles: {len(case['profiles'])}")

# Check one profile
x_loc = 0.0825000   # 0.0375000   
profile = case['profiles'][x_loc]
print(f"\nProfile at x={x_loc}:")
print(f"  z-points: {len(profile['z'])}")
print(f"  z range: [{profile['z'].min():.4f}, {profile['z'].max():.4f}] m")
print(f"  U_x range: [{profile['U_x'].min():.4f}, {profile['U_x'].max():.4f}] m/s")
```
